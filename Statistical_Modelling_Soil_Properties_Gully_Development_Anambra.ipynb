```python
# =============================================================================
# CELL 1: GITHUB REPOSITORY CONNECTION AND DATASET DISCOVERY
# =============================================================================
#
# PURPOSE
# -------
# This cell establishes a reproducible connection between the Google Colab
# analysis environment and the GitHub repository containing the laboratory
# soil datasets used in this study:
#
# "Statistical Modelling of Soil Physicochemical Properties Influencing Gully
#  Development in Contrasting Geomorphic Settings in Anambra State, Nigeria"
#
# The script:
#   1. Clones the public GitHub repository into the Colab runtime.
#   2. Verifies that the repository has been successfully accessed.
#   3. Identifies the designated "data" directory.
#   4. Searches the data directory for all CSV files.
#   5. Reports the number, names, and file sizes of the datasets discovered.
#
# REPRODUCIBILITY NOTE
# --------------------
# The raw laboratory datasets are read directly from the GitHub repository.
# This ensures that the analysis can be reproduced from the same version of
# the source data without manually uploading files into Google Colab.
#
# Repository:
# https://github.com/NNwobi-354/soil-properties-gully-development-anambra
# =============================================================================


# -----------------------------------------------------------------------------
# 1. IMPORT REQUIRED PYTHON LIBRARIES
# -----------------------------------------------------------------------------

import os
import glob
import pandas as pd


# -----------------------------------------------------------------------------
# 2. DEFINE THE GITHUB REPOSITORY AND LOCAL COLAB DIRECTORY
# -----------------------------------------------------------------------------

GITHUB_REPOSITORY = (
    "https://github.com/NNwobi-354/"
    "soil-properties-gully-development-anambra.git"
)

REPOSITORY_NAME = "soil-properties-gully-development-anambra"

# Location where the repository will be cloned inside the Colab runtime
REPOSITORY_PATH = f"/content/{REPOSITORY_NAME}"

# Expected location of the laboratory datasets
DATA_PATH = os.path.join(REPOSITORY_PATH, "data")


# -----------------------------------------------------------------------------
# 3. CLONE THE GITHUB REPOSITORY
# -----------------------------------------------------------------------------

print("=" * 80)
print("GITHUB REPOSITORY CONNECTION")
print("=" * 80)

if not os.path.exists(REPOSITORY_PATH):

    print("Cloning GitHub repository into Google Colab...")

    !git clone {GITHUB_REPOSITORY}

    print("\nRepository successfully cloned.")

else:

    print("Repository already exists in the current Colab session.")
    print("No additional cloning is required.")


# -----------------------------------------------------------------------------
# 4. VERIFY THAT THE REPOSITORY WAS SUCCESSFULLY ACCESSED
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("REPOSITORY VERIFICATION")
print("=" * 80)

if os.path.exists(REPOSITORY_PATH):

    print("✓ GitHub repository successfully accessed.")
    print(f"✓ Local repository path: {REPOSITORY_PATH}")

else:

    raise FileNotFoundError(
        "The GitHub repository could not be located. "
        "Please check the repository URL."
    )


# -----------------------------------------------------------------------------
# 5. VERIFY THE DATA DIRECTORY
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("DATA DIRECTORY VERIFICATION")
print("=" * 80)

if os.path.exists(DATA_PATH):

    print("✓ Data directory successfully located.")
    print(f"✓ Dataset directory: {DATA_PATH}")

else:

    raise FileNotFoundError(
        "The 'data' directory was not found in the GitHub repository. "
        "Please verify that the laboratory CSV files have been uploaded "
        "to the repository's data folder."
    )


# -----------------------------------------------------------------------------
# 6. IDENTIFY ALL CSV DATASETS
# -----------------------------------------------------------------------------

csv_files = sorted(
    glob.glob(os.path.join(DATA_PATH, "*.csv"))
)

print("\n" + "=" * 80)
print("LABORATORY DATASET DISCOVERY")
print("=" * 80)

print(f"Total CSV datasets identified: {len(csv_files)}")


# -----------------------------------------------------------------------------
# 7. DISPLAY DATASET INFORMATION
# -----------------------------------------------------------------------------

if len(csv_files) == 0:

    print("\nWARNING: No CSV files were found in the data directory.")

else:

    print("\nAvailable laboratory datasets:\n")

    for number, file in enumerate(csv_files, start=1):

        file_name = os.path.basename(file)
        file_size_kb = os.path.getsize(file) / 1024

        print(
            f"{number:02d}. {file_name} "
            f"({file_size_kb:.2f} KB)"
        )


# -----------------------------------------------------------------------------
# 8. CREATE A DATASET INVENTORY
# -----------------------------------------------------------------------------

dataset_inventory = pd.DataFrame({
    "Dataset_Number": range(1, len(csv_files) + 1),
    "Dataset_Name": [os.path.basename(f) for f in csv_files],
    "File_Path": csv_files,
    "File_Size_KB": [
        round(os.path.getsize(f) / 1024, 2)
        for f in csv_files
    ]
})


# -----------------------------------------------------------------------------
# 9. DISPLAY THE DATASET INVENTORY
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("DATASET INVENTORY")
print("=" * 80)

display(dataset_inventory)


# -----------------------------------------------------------------------------
# 10. FINAL REPRODUCIBILITY CHECK
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("REPRODUCIBILITY STATUS")
print("=" * 80)

print("✓ Repository connection completed.")
print("✓ Data directory verified.")
print(f"✓ {len(csv_files)} CSV dataset(s) detected.")
print("✓ Dataset inventory generated.")
print("\nThe laboratory datasets are now accessible from Google Colab.")
```
